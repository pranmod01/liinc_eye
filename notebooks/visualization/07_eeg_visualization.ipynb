# EEG Data Preprocessing and Visualization

This notebook provides a comprehensive analysis of EEG data.

**Data Structure:**
- **Teams/Subjects**: Multiple teams (Team 4, 6, 7, etc.) each with up to 3 sessions
- **Sessions**: Each session recorded on a specific date/time with data from multiple computers
- **EEG arrays**: `n_time_samples` × 20 channels per trial (submit-locked −2 s to +2 s at 256 Hz → typically 1024 × 20)
- **Recording setup**: Multiple computers (U9TEJGM, 9M4VCHG, 539136F, etc.) recording simultaneously

**Data Format:**
- `subject_date_id`: Format is `MMDD_HHMM_ComputerID` (e.g., `0831_1300_9M4VCHG`)
- `trial_id`: Trial number within each session
- `display_eeg`: EEG data array (`n_time_samples` × 20)

**Analyses Included:**
1. Data loading and structure exploration
2. Signal quality assessment
3. Time-domain visualization
4. Power spectral density (PSD) analysis
5. Frequency band power extraction
6. Topographic mapping
7. Cross-session and cross-computer comparisons
8. Statistical summaries

## 2. EEG Channel Information and Data Quality

```python
# Channel names matching chan_locs.sfp (10-20 system, 20 channels)
# Order matches the sfp file for electrode position mapping
channel_names = [
    'Fp1', 'F7', 'F8', 'T4', 'T6', 'T5', 'T3', 'Fp2', 'O1', 'P3',
    'Pz', 'F3', 'Fz', 'F4', 'C4', 'P4', 'POz', 'C3', 'Cz', 'O2'
]

# Channel regions for grouped analysis
channel_regions = {
    'Frontal': ['Fp1', 'Fp2', 'F7', 'F3', 'Fz', 'F4', 'F8'],
    'Central': ['T3', 'C3', 'Cz', 'C4', 'T4'],
    'Parietal': ['T5', 'P3', 'Pz', 'P4', 'T6'],
    'Occipital': ['O1', 'POz', 'O2']
}

# Channel indices for each region
region_indices = {region: [channel_names.index(ch) for ch in channels] 
                  for region, channels in channel_regions.items()}

print("EEG Channels by Region:")
for region, channels in channel_regions.items():
    indices = region_indices[region]
    print(f"  {region}: {', '.join(channels)} (indices: {indices})")
```

```python
# Data quality assessment
print("=" * 70)
print("DATA QUALITY ASSESSMENT")
print("=" * 70)

# Check for NaN and Inf values
nan_trials = []
inf_trials = []
amplitude_stats = []

for idx, row in eeg_df.iterrows():
    eeg = np.array(row['display_eeg'])  # Shape: (n_samples, 20)
    
    if np.isnan(eeg).any():
        nan_trials.append(idx)
    if np.isinf(eeg).any():
        inf_trials.append(idx)
    
    amplitude_stats.append({
        'min': eeg.min(),
        'max': eeg.max(),
        'mean': eeg.mean(),
        'std': eeg.std()
    })

amp_df = pd.DataFrame(amplitude_stats)

print(f"\nData Integrity:")
print(f"  Trials with NaN values: {len(nan_trials)}")
print(f"  Trials with Inf values: {len(inf_trials)}")

print(f"\nAmplitude Statistics (across all trials):")
print(f"  Global min: {amp_df['min'].min():.4f}")
print(f"  Global max: {amp_df['max'].max():.4f}")
print(f"  Mean of means: {amp_df['mean'].mean():.6f}")
print(f"  Mean of stds: {amp_df['std'].mean():.4f}")

# Identify potential outlier trials (extreme amplitudes)
amp_threshold = amp_df['max'].quantile(0.99)
outlier_trials = amp_df[amp_df['max'] > amp_threshold].index.tolist()
print(f"\nPotential outlier trials (amplitude > 99th percentile): {len(outlier_trials)}")
```

## 5. Power Spectral Density (PSD) Analysis

```python
# Define frequency bands
freq_bands = {
    'Delta': (0.5, 4),
    'Theta': (4, 8),
    'Alpha': (8, 13),
    'Beta': (13, 30),
    'Gamma': (30, 50)
}

def compute_psd(eeg_data, fs=256):
    """
    Compute Power Spectral Density for all channels.
    
    Args:
        eeg_data: (n_channels, n_timepoints)
        fs: sampling rate in Hz
    
    Returns:
        freqs: frequency array
        psd: power spectral density (n_channels, n_freqs)
    """
    freqs, psd = signal.welch(eeg_data, fs=fs, nperseg=min(256, eeg_data.shape[1]))
    return freqs, psd

# Compute PSD for current trial
freqs, psd = compute_psd(eeg_data, fs)

print(f"Frequency range: {freqs[0]:.2f} - {freqs[-1]:.2f} Hz")
print(f"Frequency resolution: {freqs[1] - freqs[0]:.2f} Hz")
print(f"PSD shape: {psd.shape} (channels x frequency bins)")
```

```python
# Plot PSD for each region (averaged across channels)
fig, axes = plt.subplots(2, 2, figsize=(14, 10))
axes = axes.flatten()

colors = plt.cm.Set2(np.linspace(0, 1, len(freq_bands)))
band_colors = dict(zip(freq_bands.keys(), colors))

for idx, (region, channels) in enumerate(channel_regions.items()):
    ax = axes[idx]
    
    # Get channel indices for this region
    ch_indices = [channel_names.index(ch) for ch in channels]
    
    # Average PSD across channels in this region
    psd_region = np.mean(psd[ch_indices, :], axis=0)
    
    # Plot PSD
    ax.semilogy(freqs, psd_region, 'k-', linewidth=2, label='PSD')
    
    # Mark frequency bands
    for band_name, (f_low, f_high) in freq_bands.items():
        mask = (freqs >= f_low) & (freqs <= f_high)
        ax.fill_between(freqs[mask], psd_region[mask], alpha=0.3, 
                       color=band_colors[band_name], label=band_name)
    
    ax.set_xlabel('Frequency (Hz)')
    ax.set_ylabel('Power Spectral Density')
    ax.set_title(f'{region} Region')
    ax.set_xlim([0, 50])
    ax.legend(fontsize=8, loc='upper right')
    ax.grid(alpha=0.3)

plt.suptitle(f'Power Spectral Density by Region - Trial {trial_idx}', fontsize=14, y=0.995)
plt.tight_layout()
plt.show()
```

## 6. Extract Band Power Features

```python
def compute_band_power(eeg_data, fs=256, bands=None):
    """
    Compute band power for each channel and frequency band.
    
    Args:
        eeg_data: (n_channels, n_timepoints)
        fs: sampling rate
        bands: dict of {band_name: (f_low, f_high)}
    
    Returns:
        band_powers: dict of {band_name: array of shape (n_channels,)}
    """
    if bands is None:
        bands = freq_bands
    
    freqs, psd = compute_psd(eeg_data, fs)
    
    band_powers = {}
    for band_name, (f_low, f_high) in bands.items():
        # Find frequency indices
        freq_mask = (freqs >= f_low) & (freqs <= f_high)
        
        # Integrate power in this band
        band_power = np.trapz(psd[:, freq_mask], freqs[freq_mask], axis=1)
        band_powers[band_name] = band_power
    
    return band_powers

# Compute band powers for current trial
bp = compute_band_power(eeg_data, fs)

print("Band Powers (current trial):")
print("-" * 50)
for band_name, powers in bp.items():
    print(f"  {band_name:8s}: mean={np.mean(powers):.3e}, std={np.std(powers):.3e}")
```

```python
# Visualize band powers across channels
fig, ax = plt.subplots(figsize=(14, 6))

# Create band power matrix
band_power_matrix = np.array([bp[band] for band in freq_bands.keys()])

# Normalize for better visualization (log scale)
band_power_matrix_log = np.log10(band_power_matrix + 1e-10)

im = ax.imshow(band_power_matrix_log, aspect='auto', cmap='viridis')
ax.set_yticks(range(len(freq_bands)))
ax.set_yticklabels(freq_bands.keys())
ax.set_xticks(range(len(channel_names)))
ax.set_xticklabels(channel_names, rotation=45, ha='right')
ax.set_title(f'Band Power Across Channels (log scale) - Trial {trial_idx}')
plt.colorbar(im, ax=ax, label='log10(Power)')

plt.tight_layout()
plt.show()
```

## 7. Grand Average Analysis Across All Trials

```python
# Compute grand average EEG and band powers across all trials
print("Computing grand average across all trials...")
print("(This may take a moment for large datasets)")

# Sample a subset for faster computation
sample_size = min(1000, len(eeg_df))
sample_indices = np.random.choice(len(eeg_df), sample_size, replace=False)

all_band_powers = {band: [] for band in freq_bands.keys()}
all_eeg = []

for idx in sample_indices:
    eeg = np.array(eeg_df['display_eeg'].iloc[idx]).T  # (20, n_samples)
    all_eeg.append(eeg)
    
    bp_trial = compute_band_power(eeg, fs)
    for band in freq_bands.keys():
        all_band_powers[band].append(bp_trial[band])

# Compute grand averages
grand_avg_eeg = np.mean(all_eeg, axis=0)  # (20, n_samples)
grand_avg_bp = {band: np.mean(powers, axis=0) for band, powers in all_band_powers.items()}
grand_std_bp = {band: np.std(powers, axis=0) for band, powers in all_band_powers.items()}

print(f"\nGrand Average Band Powers (n={sample_size} trials):")
print("-" * 60)
for band in freq_bands.keys():
    mean_power = np.mean(grand_avg_bp[band])
    std_power = np.mean(grand_std_bp[band])
    print(f"  {band:8s}: mean={mean_power:.3e} +/- {std_power:.3e}")
```

```python
# Visualize grand average EEG waveform
fig, axes = plt.subplots(4, 1, figsize=(15, 12), sharex=True)

for idx, (region, channels) in enumerate(channel_regions.items()):
    ax = axes[idx]
    
    for ch_name in channels:
        ch_idx = channel_names.index(ch_name)
        ax.plot(time, grand_avg_eeg[ch_idx, :], alpha=0.7, label=ch_name, linewidth=1)
    
    ax.set_ylabel('Amplitude')
    ax.set_title(f'{region} Channels - Grand Average')
    ax.legend(loc='upper right', ncol=len(channels), fontsize=8)
    ax.grid(alpha=0.3)

axes[-1].set_xlabel('Time (s)')
plt.suptitle(f'Grand Average EEG (n={sample_size} trials)', fontsize=14, y=0.995)
plt.tight_layout()
plt.show()
```

## 8. Topographic Maps

```python
# Electrode positions from chan_locs.sfp file (10-20 system)
# Coordinates are (x, y) in 2D projection for topographic mapping
electrode_positions = {
    'Fp1': (-0.30883, 0.95048), 'Fp2': (0.30883, 0.95048),
    'F7': (-0.80852, 0.58743), 'F3': (-0.54501, 0.67303), 'Fz': (0, 0.71934), 'F4': (0.54501, 0.67303), 'F8': (0.80852, 0.58743),
    'T3': (-0.99939, 0), 'C3': (-0.71934, 0), 'Cz': (0, 0), 'C4': (0.71934, 0), 'T4': (0.99939, 0),
    'T5': (-0.80852, -0.58743), 'P3': (-0.54501, -0.67303), 'Pz': (0, -0.71934), 'P4': (0.54501, -0.67303), 'T6': (0.80852, -0.58743),
    'O1': (-0.30883, -0.95048), 'POz': (0, -0.93358), 'O2': (0.30883, -0.95048)
}

# Extract positions in order
positions = np.array([electrode_positions[ch] for ch in channel_names])

print("Electrode positions loaded from chan_locs.sfp for topographic plotting")
```

```python
# Plot topographic maps for each frequency band (grand average)
fig, axes = plt.subplots(1, 5, figsize=(20, 4))

# Create interpolation grid
xi = np.linspace(-1, 1, 100)
yi = np.linspace(-1, 1, 100)
xi, yi = np.meshgrid(xi, yi)

# Use log-transformed band powers for better visualization
for idx, (band_name, powers) in enumerate(grand_avg_bp.items()):
    ax = axes[idx]
    
    # Log transform
    powers_log = np.log10(powers + 1e-10)
    
    # Interpolate
    zi = griddata(positions, powers_log, (xi, yi), method='cubic', fill_value=np.nan)
    
    # Mask outside head
    mask = xi**2 + yi**2 > 1
    zi[mask] = np.nan
    
    # Plot
    im = ax.contourf(xi, yi, zi, levels=15, cmap='viridis')
    ax.scatter(positions[:, 0], positions[:, 1], c='white', s=30, zorder=10, edgecolors='black')
    
    # Add channel labels
    for ch_idx, ch_name in enumerate(channel_names):
        ax.annotate(ch_name, positions[ch_idx], fontsize=6, ha='center', va='center', color='white')
    
    # Draw head outline
    circle = plt.Circle((0, 0), 1, fill=False, color='black', linewidth=2)
    ax.add_patch(circle)
    # Draw nose
    ax.plot([0, 0], [1, 1.15], 'k-', linewidth=2)
    
    ax.set_xlim([-1.2, 1.2])
    ax.set_ylim([-1.2, 1.25])
    ax.set_aspect('equal')
    ax.axis('off')
    ax.set_title(f'{band_name}\n({freq_bands[band_name][0]}-{freq_bands[band_name][1]} Hz)', fontsize=11)
    
    plt.colorbar(im, ax=ax, fraction=0.046, pad=0.04, label='log10(Power)')

plt.suptitle(f'Topographic Maps - Grand Average Band Power (n={sample_size} trials)', fontsize=14, y=1.08)
plt.tight_layout()
plt.show()
```

## 9. Session-Level Statistics and Comparisons

```python
# Compute session-level statistics
print("Computing session-level band power statistics...")

session_stats = []
for session in eeg_df['session_datetime'].unique():
    session_df = eeg_df[eeg_df['session_datetime'] == session]
    
    # Sample up to 50 trials per session
    sample_df = session_df.sample(min(50, len(session_df)))
    
    session_bp = {band: [] for band in freq_bands.keys()}
    for _, row in sample_df.iterrows():
        eeg = np.array(row['display_eeg']).T
        bp_trial = compute_band_power(eeg, fs)
        for band in freq_bands.keys():
            session_bp[band].append(np.mean(bp_trial[band]))  # Average across channels
    
    stats_row = {
        'session': session,
        'n_trials': len(session_df),
        'computers': list(session_df['computer_id'].unique())
    }
    for band in freq_bands.keys():
        stats_row[f'{band}_mean'] = np.mean(session_bp[band])
        stats_row[f'{band}_std'] = np.std(session_bp[band])
    
    session_stats.append(stats_row)

session_stats_df = pd.DataFrame(session_stats)
print(f"\nSession-level statistics computed for {len(session_stats_df)} sessions")
session_stats_df.head(10)
```

```python
# Visualize session-level band power variability
fig, axes = plt.subplots(2, 3, figsize=(15, 10))
axes = axes.flatten()

for idx, band in enumerate(freq_bands.keys()):
    ax = axes[idx]
    
    # Box plot of band power across sessions
    data = session_stats_df[f'{band}_mean'].values
    ax.boxplot(data, vert=True)
    ax.scatter(np.ones(len(data)) + np.random.randn(len(data))*0.04, data, 
               alpha=0.5, color='steelblue', s=30)
    
    ax.set_ylabel('Mean Band Power')
    ax.set_title(f'{band} Band ({freq_bands[band][0]}-{freq_bands[band][1]} Hz)')
    ax.set_xticklabels(['All Sessions'])
    ax.grid(alpha=0.3, axis='y')

# Remove extra subplot
axes[-1].axis('off')

plt.suptitle('Session-Level Band Power Variability', fontsize=14, y=1.01)
plt.tight_layout()
plt.show()

# Print summary statistics
print("\n" + "=" * 70)
print("BAND POWER SUMMARY ACROSS SESSIONS")
print("=" * 70)
for band in freq_bands.keys():
    mean_col = f'{band}_mean'
    print(f"\n{band} Band:")
    print(f"  Overall mean: {session_stats_df[mean_col].mean():.4e}")
    print(f"  Std across sessions: {session_stats_df[mean_col].std():.4e}")
    print(f"  Min session: {session_stats_df[mean_col].min():.4e}")
    print(f"  Max session: {session_stats_df[mean_col].max():.4e}")
```

## 10. Computer-Level Comparison

```python
# Compare band power across different computers
print("Computing computer-level statistics...")

computer_stats = []
for computer in eeg_df['computer_id'].unique():
    comp_df = eeg_df[eeg_df['computer_id'] == computer]
    
    # Sample trials
    sample_df = comp_df.sample(min(200, len(comp_df)))
    
    comp_bp = {band: [] for band in freq_bands.keys()}
    for _, row in sample_df.iterrows():
        eeg = np.array(row['display_eeg']).T
        bp_trial = compute_band_power(eeg, fs)
        for band in freq_bands.keys():
            comp_bp[band].append(np.mean(bp_trial[band]))
    
    stats_row = {'computer': computer, 'n_trials': len(comp_df)}
    for band in freq_bands.keys():
        stats_row[f'{band}_mean'] = np.mean(comp_bp[band])
        stats_row[f'{band}_std'] = np.std(comp_bp[band])
    
    computer_stats.append(stats_row)

computer_stats_df = pd.DataFrame(computer_stats)

# Visualize
fig, axes = plt.subplots(2, 3, figsize=(15, 10))
axes = axes.flatten()

computers = computer_stats_df['computer'].tolist()
x = np.arange(len(computers))

for idx, band in enumerate(freq_bands.keys()):
    ax = axes[idx]
    
    means = computer_stats_df[f'{band}_mean'].values
    stds = computer_stats_df[f'{band}_std'].values
    
    bars = ax.bar(x, means, yerr=stds, capsize=5, color=plt.cm.Set2(idx/len(freq_bands)), 
                  edgecolor='black', alpha=0.8)
    
    ax.set_xticks(x)
    ax.set_xticklabels(computers, rotation=45, ha='right')
    ax.set_ylabel('Mean Band Power')
    ax.set_title(f'{band} Band')
    ax.grid(alpha=0.3, axis='y')

axes[-1].axis('off')

plt.suptitle('Band Power Comparison Across Computers', fontsize=14, y=1.01)
plt.tight_layout()
plt.show()

print("\nComputer Statistics:")
print(computer_stats_df.to_string())
```
